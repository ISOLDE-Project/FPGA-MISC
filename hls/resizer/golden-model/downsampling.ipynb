```python
import torch
import torch.nn as nn
import torchvision.transforms as T
from PIL import Image

# Load image as grayscale
#testInput="testpattern-hd-1080.png"
testInput="grey_design.png"
img = Image.open(testInput).convert("L")  # Load as grayscale
img_tensor = T.ToTensor()(img).unsqueeze(0)  # shape [1, 1, H, W]
print( f"input shape: {img_tensor.shape}")



# Define mean filter convolution layer
channels = 1  # grayscale
kernel_size = 5
stride = (5, 5)
padding = 0  
#
VGA_H = 224
VGA_W = 224

conv = nn.Conv2d(
    in_channels=channels,
    out_channels=1,
    kernel_size=kernel_size,
    stride=stride,
    padding=padding,
    bias=False
)

# Set weights to mean filter
with torch.no_grad():
    mean_kernel = torch.ones((1, 1, kernel_size, kernel_size)) / (kernel_size ** 2)
    conv.weight.copy_(mean_kernel)

# Apply convolution
downsampled = conv(img_tensor)  
print( f"downsampled shape: {downsampled.shape}")
x_start = (downsampled.shape[2]-VGA_H)//2
y_start = (downsampled.shape[3]-VGA_W)//2

# Save output
output = downsampled[:,:,x_start:x_start+VGA_H,y_start:y_start+VGA_W]
print( f"output shape: {output.shape}")
output_img = T.ToPILImage()(output.squeeze(0).clamp(0, 1))
output_img.save("output_downsampled.png")


```
